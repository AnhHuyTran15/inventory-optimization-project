## Situation

A retail network operates 10 stores x 50 SKUs (500 store-item pairs). Daily demand history (2013-2017) and real GPS shipment-tracking history (lead times, vehicle types, coordinates) both exist but have never been combined to drive a replenishment decision.

## Complication

A static EOQ / fixed reorder point treats demand and lead time as known constants. Measured data shows otherwise: demand CV averages ~0.55 across SKUs, and lead time ranges from <1 day to ~18 days depending on vehicle type. A closed-form formula cannot represent this, cannot exploit imperfect correlation between stores (risk pooling), and ignores that replenishment requires a truck with finite capacity.

## Question

Given *measured* demand and lead-time variability, what replenishment policy - how much safety stock, held where, delivered by which routes - minimizes total system cost at a target service level?
