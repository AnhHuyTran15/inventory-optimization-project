## Formula

**Decentralized:** $SS = z\sqrt{L}\sum_i \sigma_i$

**Pooled:** $SS = z\sqrt{L}\sqrt{\sum_i \sigma_i^2 + \sum_{i \ne j} \rho_{ij}\sigma_i\sigma_j}$

Pooled $\le$ Decentralized whenever $\rho_{ij} < 1$ — this is the risk-pooling effect, quantified rather than asserted.

```python
import matplotlib.pyplot as plt

plt.scatter(pooling_results['mean_pairwise_correlation'], pooling_results['pct_savings'])
plt.xlabel('Mean pairwise correlation across stores')
plt.ylabel('% safety stock saved by pooling')
plt.title('Sanity check: lower correlation -> higher pooling savings')
plt.show()
```
